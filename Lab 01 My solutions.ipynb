## My solution for problem 2

```python
# Example: Zooming in on a range near x = -7
plt.figure(figsize=(10, 7))
plt.plot(x, y)
plt.xlim(4.1, 3.9) # Zoom x-axis to show area around x = -7
plt.ylim(0, 2) # Zoom y-axis to show area around y = 1
# Shows the y=1 line
plt.axhline(y=1, color='r', linestyle='--')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Plus Point')
plt.grid(True)
plt.show()


# Graph 2 (Negative Graph)
plt.figure(figsize=(10, 7))
plt.plot(x, y)
plt.xlim(-2.7, -2.5) # Zoom x-axis to show area around x = -7
plt.ylim(0, 2) # Zoom y-axis to show area around y = 1
# Shows the y=1 line
plt.axhline(y=1, color='r', linestyle='--')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Negative Point')
plt.grid(True)
plt.show()
```
